Shallow Copy and Deep Copy

```python
a = [10, 20, 30]
print(a)
```

```python
b=a
print(b)
```

```python
b.append(40)
print(a)
print(b)
```

kio ye ?
kio k a aur b ke paas same object ka reference hai

```python
b = a.copy()

b.append(40)

print(a)
print(b)
```

a → [10, 20, 30]

b → [10, 20, 30, 40]

Ye shallow copy hai.
